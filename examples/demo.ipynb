$$F_0=0,\quad F_1=1,\quad F_n=F_{n-1}+F_{n-2}$$

```python
def fibonacci_recursive(n: int) -> int:
    if n < 0:
        raise ValueError("n must be non-negative")
    if n < 2:
        return n
    return fibonacci_recursive(n - 1) + fibonacci_recursive(n - 2)
```

$$(F_k,F_{k+1})\leftarrow(F_{k+1},F_k+F_{k+1})$$

```python
def fibonacci_iterative(n: int) -> int:
    if n < 0:
        raise ValueError("n must be non-negative")

    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a
```

$$F_{2k}=F_k(2F_{k+1}-F_k),\qquad F_{2k+1}=F_k^2+F_{k+1}^2$$

```python
def fibonacci_fast_doubling(n: int) -> int:
    if n < 0:
        raise ValueError("n must be non-negative")

    def fib_pair(k: int) -> tuple[int, int]:
        if k == 0:
            return 0, 1

        a, b = fib_pair(k // 2)
        c = a * (2 * b - a)
        d = a * a + b * b

        return (c, d) if k % 2 == 0 else (d, c + d)

    return fib_pair(n)[0]
```
