$$F_0=0,\quad F_1=1,\quad F_n=F_{n-1}+F_{n-2}\;(n\ge2)$$

```python
def fib_recursive(n):
    """Return F_n using direct recursion: O(2^n) time, O(n) space."""
    if n < 0:
        raise ValueError("n must be non-negative")
    if n < 2:
        return n
    return fib_recursive(n - 1) + fib_recursive(n - 2)
```

$$(F_k,F_{k+1})\leftarrow(F_{k+1},F_k+F_{k+1}),\qquad (F_0,F_1)=(0,1)$$

```python
def fib_iterative(n):
    """Return F_n iteratively: O(n) time, O(1) space."""
    if n < 0:
        raise ValueError("n must be non-negative")
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a
```

$$
F_{2k}=F_k(2F_{k+1}-F_k)
\quad
F_{2k+1}=F_k^2+F_{k+1}^2
$$

```python
def fib_fast_doubling(n):
    """Return F_n using fast doubling: O(log n) arithmetic steps."""
    if n < 0:
        raise ValueError("n must be non-negative")

    def pair(k):
        if k == 0:
            return 0, 1
        a, b = pair(k // 2)
        c = a * (2 * b - a)
        d = a * a + b * b
        return (d, c + d) if k % 2 else (c, d)

    return pair(n)[0]
```
