```python
import random
data = [random.gauss(0, 1) for _ in range(1000)]
```

```python
%%ask
Tell me whether data looks normal or not. Usually what quantitative tests are done for this?
Run whatever cells you need, then give me the verdict with the numbers you measured.
```

```python
import numpy as np
from scipy import stats

x = np.asarray(data)
n = len(x)

mean = x.mean()
std = x.std(ddof=1)
skew = stats.skew(x, bias=False)
excess_kurtosis = stats.kurtosis(x, fisher=True, bias=False)

shapiro_stat, shapiro_p = stats.shapiro(x)
dagostino_stat, dagostino_p = stats.normaltest(x)
anderson = stats.anderson(x, dist="norm")

print(f"n = {n}")
print(f"mean = {mean:.6f}")
print(f"sample std = {std:.6f}")
print(f"skewness = {skew:.6f}")
print(f"excess kurtosis = {excess_kurtosis:.6f}")
print(f"Shapiro-Wilk: W = {shapiro_stat:.6f}, p = {shapiro_p:.6f}")
print(f"D'Agostino-Pearson: K^2 = {dagostino_stat:.6f}, p = {dagostino_p:.6f}")
print(f"Anderson-Darling: A^2 = {anderson.statistic:.6f}")
print("Anderson-Darling critical values:")
for level, critical in zip(anderson.significance_level, anderson.critical_values):
    verdict = "reject normality" if anderson.statistic > critical else "do not reject normality"
    print(f"  alpha = {level/100:.3f}: critical = {critical:.3f} -> {verdict}")
```
